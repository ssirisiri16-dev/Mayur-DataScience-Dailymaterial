```python
radius=float(input("enter the radius"))
p_x=int(input("enter x_cordinate of point p"))
p_y=int(input("enter y_cordinate of point p"))
c_x=0
c_y=0
d=(((p_y-c_y)**2)+((p_x-c_x)**2))**0.5
if d<radius:
    print("point P lies inside the circle")
if d>radius:
    print("point P lies outside the circle")
if d==radius:
    print("point P lies on the circle")

```

```python
x1=int(input("enter x_component of vector 1"))
y1=int(input("enter y_component of vector 1"))
x2=int(input("enter x_component of vector 2"))
y2=int(input("enter y_component of vector 2"))
determinant=(x1*(y2-1)-y1*(x2-1)+1*(x2-y2))
if determinant==0:
    print("vectors are collinear")
else:
    print("vectors are not collinear")
       
```

```python
n=int(input("enter any number"))
k=int(input("enter kth bit"))
if n&(1<<k)!=0:
    print("kth bit is set")
else:
    print("kth bit is not set")
```
