```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("diet_recommendation_dataset_1000.csv")
```

```python
df
```

```python
def creating_categorical(x):
    if x<=30:
        return "young"
    elif x>=31 and x<=45:
        return "adult"
    else:
        return "senior"
```

```python
df["age_category"]=df["Age"].apply(creating_categorical)
```

```python
df
```

```python
young=df[df["age_category"]=="young"]["Height_cm"]
```

```python
young
```

```python
adult=df[df["age_category"]=="adult"]["Height_cm"]
```

```python
senior=df[df["age_category"]=="senior"]["Height_cm"]
```

```python
young_mean=np.mean(young)
```

```python
adult_mean=np.mean(adult)
```

```python
senior_mean=np.mean(senior)
```

```python
overall_mean=np.mean(df["Height_cm"])
```

```python
young_array=np.array(young)
```

```python
adult_array=np.array(adult)
```

```python
senior_array=np.array(senior)
```

```python
#SSB=summation ni*(xi-bar- x-bar)**2
#SSB=smmaton ni*(group_mean-overall_mean)**2
SSB=(len(young_array)*(young_mean-overall_mean)**2)+(len(adult_array)*(adult_mean-overall_mean)**2)+(len(senior_array)*(senior_mean-overall_mean)**2)
```

```python
SSB
```

```python
#SSW=summation(x-x-bar)**2
SSW=np.sum((young_array-overall_mean)**2)+np.sum((adult_array-overall_mean)**2)+np.sum((senior_array-overall_mean)**2)
```

```python
SSW
```

```python
# Df_between=k-1
#Df_within=N-k
Df_between=3-1
Df_within=len(df["Height_cm"])-3
```

```python
#mean squares
#MSB=SSB/df_between
#MSW=SSW/ Df_within 
#F_statistics=MSB/MSW
MSB=SSB/Df_between
```

```python
MSW=SSW/ Df_within
```

```python
F_statistics=MSB/MSW
```
