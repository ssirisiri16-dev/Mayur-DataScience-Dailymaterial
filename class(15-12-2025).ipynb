```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("laptopdata.csv")
```

```python
df["Cpu"].unique()
```

```python
def conver_cpu(x):
    lst=x.split()
    first_three=lst[0:3]
    return " ".join(first_three)
    
```

```python
df["Cpu_name"]=df["Cpu"].apply(conver_cpu)
```

```python
df.head(3)
```

```python
df["Cpu_name"].value_counts()
```

```python
df["Cpu_name"].unique()
```

```python
def convert_to_word(x):
    if "Intel" in x:
        return "Intel"
    elif "AMD" in x:
        return "AMD"
    else:
        return "Other"
```

```python
df["Cpu_cat"]=df["Cpu_name"].apply(convert_to_word)
```

```python
df.head(3)
```

```python
df["Cpu_cat"].unique()
```

```python
df.drop(columns=["Cpu","Cpu_name"],inplace=True)
```

```python
df["Memory"]=df["Memory"].str.replace(".0","")
```

```python
df["Memory"]=df["Memory"].str.replace("GB","")
df["Memory"]=df["Memory"].str.replace("TB","000")
```

```python
df["tp1"]=new[0]
df["tp2"]=new[1]
```

```python
df["HDD1"]=df["tp1"].apply(lambda x:1 if "HDD" in x else 0)
df["SSD1"]=df["tp1"].apply(lambda x:1 if "SSD" in x else 0)
df["Hybrid1"]=df["tp1"].apply(lambda x:1 if "Hybrid" in x else 0)
df["Flash1"]=df["tp1"].apply(lambda x:1 if "Flash Storage" in x else 0)
```

```python
df["tp2"].fillna("0",inplace=True)
```

```python
df["HDD2"]=df["tp2"].apply(lambda x:1 if "HDD" in x else 0)
df["SSD2"]=df["tp2"].apply(lambda x:1 if "SSD" in x else 0)
df["Hybrid2"]=df["tp2"].apply(lambda x:1 if "Hybrid" in x else 0)
df["Flash2"]=df["tp2"].apply(lambda x:1 if "Flash Storage" in x else 0)
```

```python
df["tp1"]=df["tp1"].str.replace("\D","",regex=True)
```

```python
df["HDD"]=((df["HDD1"]*df["tp1"])+(df["HDD2"]*df["tp2"]))
df["SSD"]=((df["SSD1"]*df["tp1"])+(df["SSD2"]*df["tp2"]))
df["Hybrid"]=((df["Hybrid1"]*df["tp1"])+(df["Hybrid2"]*df["tp2"]))
df["Flash"]=((df["Flash1"]*df["tp1"])+(df["Flash2"]*df["tp2"]))
```

```python
def convert_gpu(x):
    words=x.split()
    first_three=words[0:3]
    return " ".join(first_three)
```

```python
df["Gpu_name"]=df["Gpu"].apply(convert_gpu)
```

```python
def gpu_category(x):
    if "Intel" in x:
        return "Intel"
    elif "Nvidia" in x:
        return "Nvidia"
    elif "AMD" in x:
        return "AMD"
    else:
        return "other"
```

```python
df["Gpu_cat"]=df["Gpu_name"].apply(gpu_category)
```
