```python
def check_subset(s1,s2):
    for i in s1:
        if i not in s2:
            return False
    return True
s1={1,7,8}
s2={1,7,8,9}
check_subset(s1,s2)    
```

```python
def check_isdisjoint(s1,s2):
    for i in s1:
        if i in s2:
            return False
    return True
s1={1,7,8}
s2={1, 7, 8, 9}
check_isdisjoint(s1,s2)
```

```python
def my_pop(dict,key):
    new_dict={}
    for i,j in dict.items():
        if i!=key:
            new_dict[i]=j
    return new_dict
dict={1: 100, 3: 300} 
my_pop(dict,1)
```
