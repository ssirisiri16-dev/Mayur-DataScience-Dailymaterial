```python
def checking_ou_of_3rd_range(data):
    count=0
    mean=np.mean(data)
    sd=np.std(data)
    LL=mean-3*sd 
    UL=mean+3*sd
    for i in data:
        if i>UL or i<LL:
            count+=1
    return count
```

```python
result=checking_ou_of_3rd_range(df["Height_cm"])
```

```python
result
```

```python
def outlier_detection_quantile_method(column):
    count=0
    LL1=np.quantile(column,0.05)
    UL1=np.quantile(column,0.95)
    for i in column:
        if i>UL1 or i<LL1:
            count+=1
    return count
```

```python
result=outlier_detection_quantile_method(df["Cholesterol"])
```
