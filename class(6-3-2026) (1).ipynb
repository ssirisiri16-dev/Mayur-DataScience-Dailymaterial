```python
X=df[["petal_length","petal_width"]]
```

```python
Y=df["species"]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
ss=StandardScaler()
```

```python
le=LabelEncoder()
```

```python
xtrain_scaled=ss.fit_transform(xtrain)
```

```python
xtest_scaled=ss.fit_transform(xtest)
```

```python
ytrain_le=le.fit_transform(ytrain)
```

```python
ytest_le=le.fit_transform(ytest)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
param_grid={
    "n_neighbors":[3,5,7,9,11,13,15],
    "weights":["uniform","distance"],
    "algorithm":["ball_tree", "kd_tree", "brute"],
    "metric":["euclidean","manhattan"]
}
```

```python
grid=GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)
```

```python
grid.fit(xtrain_scaled,ytrain_le)
```

```python
grid.best_params_
```

```python
model1=KNeighborsClassifier(algorithm='ball_tree',metric='manhattan',n_neighbors= 5,weights= 'uniform')
```

```python
model1.fit(xtrain_scaled,ytrain_le)
```

```python
from sklearn.metrics import accuracy_score
```

```python
ypred=model1.predict(xtest_scaled)
```

```python
acc=accuracy_score(ytest_le,ypred)
```
