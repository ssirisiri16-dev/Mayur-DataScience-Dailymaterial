```python
dff=df[["Survived","Pclass"]].dropna()
```

```python
dff
```

```python
survived=[
    len(dff[(dff["Survived"]==0) & (dff["Pclass"]==1)]),
    len(dff[(dff["Survived"]==0) & (dff["Pclass"]==2)]),
    len(dff[(dff["Survived"]==0) & (dff["Pclass"]==3)])
]
```

```python
survived
```

```python
unsurvived=[
    len(dff[(dff["Survived"]==1) & (dff["Pclass"]==1)]),
    len(dff[(dff["Survived"]==1) & (dff["Pclass"]==2)]),
    len(dff[(dff["Survived"]==1) & (dff["Pclass"]==3)])
]
```

```python
unsurvived
```

```python
observed=np.array([survived,unsurvived])
```

```python
observed
```

```python
row_sum=np.sum(observed,axis=1)
```

```python
row_sum
```

```python
col_sum=np.sum(observed,axis=0)
```

```python
col_sum
```

```python
total_sum=np.sum(observed)
```

```python
total_sum
```

```python
expected=np.zeros(observed.shape)
```

```python
expected
```

```python
for i in range(observed.shape[0]):
    for j in range(observed.shape[1]):
        expected[i,j]=(row_sum[i]*col_sum[j])/total_sum
        
```

```python
expected
```

```python
chi2_value=np.sum(((observed-expected)**2)/expected)
```

```python
chi2_value
```
