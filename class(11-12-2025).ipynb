```python
data1=df1[["id","season"]]
data2=df2[["match_id","bowler","over","total_runs"]] 
res=pd.merge(data1,data2,left_on="id",right_on="match_id",how="inner") 
bowler=res.groupby(["season","bowler"])["total_runs"].sum().reset_index(name="sum") 
balls_count = res.groupby(["season", "bowler"])["over"].count().reset_index(name="balls_bowled")                                                                                
balls_count["overs_bowled"] = balls_count["balls_bowled"] / 6 
bowler_stats=pd.merge(bowler,balls_count,on=["season","bowler"]) 
bowler_stats["economy"]=bowler_stats["sum"]/bowler_stats["overs_bowled"]
```

```python
bowler_stats
```

```python
res.groupby(["season","bowler"])["total_runs"].sum()/pd.Series(["over"]).count()
```
