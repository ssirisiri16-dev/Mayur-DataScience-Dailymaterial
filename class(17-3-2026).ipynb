```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss=StandardScaler()
```

```python
X_scaled=ss.fit_transform(X)
```

```python
X_scaled
```

```python
lof=LocalOutlierFactor(n_neighbors=20,contamination=0.05)
```

```python
lof_labels=lof.fit_predict(X_scaled)
```

```python
lof_labels
```

```python
df["LOF"]=lof_labels
```

```python
df
```

```python
df["LOF"].value_counts()
```

```python
df.columns
```

```python
def LOF_outlier_imputation(df,cols,low=0.05,high=0.95):
    inliner=df[df["LOF"]==1]
    outliers=df[df["LOF"]==-1]
    for col in cols:
        LL=inliner[col].quantile(low)
        UL=inliner[col].quantile(high)
        outliers[col]=outliers[col].apply(lambda x:LL if x<=LL else (UL if UL>x else x))
    new_data=pd.concat([inliner,outliers])
    return new_data    
```

```python
cols=["Age","Height_cm","Weight_kg","BMI","Sugar_Level","Cholesterol"]
new_data=LOF_outlier_imputation(df,cols)
```
