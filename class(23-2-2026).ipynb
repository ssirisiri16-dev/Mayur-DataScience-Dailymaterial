```python
import numpy as np
```

```python
points=np.random.randn(10,4)
```

```python
points.shape[1]
```

```python
def get_lst_dict(points,no_of_hyperplanes):
    dim=points.shape[1]
    hyperplanes=np.random.normal(size=(no_of_hyperplanes,dim))
    lsh_dict={}
    for i in range(points.shape[0]):
        point=points[i]
        key=[]
        for j in range(no_of_hyperplanes):
            wTx=np.dot(hyperplanes[j],point)
            if wTx>=0:
                key.append(1)
            else:
                key.append(-1)
        key=tuple(key)
        if key not in lsh_dict:
            lsh_dict[key]=[]
        lsh_dict[key].append(point)
    return lsh_dict
  
```

```python
points=np.random.randn(10,3)
no_of_hyperplanes=3
lsh_dict=get_lst_dict(points,no_of_hyperplanes)
```
